### String Indexing

```python
str1 = str(input("Enter a string of characters:"))

print(str1[2], "is the 3rd character in the string") #3rd character of string
print(str1[-4],"is the fourth last character in the string") #4th last character of string
print(str1[:3],"are the first 3 characters in the string") #first 3 characters of string 
print(str1[::2],"--> every second character of string") #every second character of string
print(str1[-3:],"--> last 3 characters of string ") #last 3 characters of string
#catch
print(str1[::-1],"reverse of the string") #reverse of string
```

#### String Slicing

```python
course_code = "24t2cs1002" # 24 - year, t2 - term 2, cs1002 - course id

course_term = int(course_code[3]) # int: get the term of the year as number from course_code
course_year = int(course_code[:2])# int: get the year as two digit number from course_code

print(course_term, course_year, course_code)
```
