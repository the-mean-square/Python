### Arithmetics

#### Sum of 2 numbers

```python
num1 = int(input("Enter the first number: "))
num2 = int(input("Enter the second number: "))
sum = num1 + num2
print("The sum is: ", sum)
```

#### Twice the sum 

```python
num1 = int(input("Enter the first number: "))
num2 = int(input("Enter the second number: "))
twice_sum = 2*(num1 + num2) # not 2(num1 + num2) 
print("Twice sum is: ", twice_sum)
```

#### Absolute difference

```python
num1 = int(input("Enter the first number: "))
num2 = int(input("Enter the second number: "))
abs_diff = abs(num1 - num2) 
print("Absolute difference is: ", abs_diff)
```

#### Absolute difference between sum and product

```python
num1 = int(input("Enter the first number: "))
num2 = int(input("Enter the second number: "))
abs_diff_sp = abs((num1 + num2) - (num1*num2)) 
print("Absolute difference is: ", abs_diff_sp)
```

### Discounts

#### Discounted price

```python
#Test case: 100, 55
price = int(input("Enter the price (type integer)"))
discount_percent = float(input("Enter the discount '%' in decimal"))

discounted_price = price*(1-discount_percent/100)
print("Discounted Price is: ", discounted_price)
```

#### Rounded discount price

```python
price = int(input("Enter the price (type integer)"))
discount_percent = float(input("Enter the discount '%' in decimal"))

rounded_discounted_price = round(price*(1-discount_percent/100))
print("Discounted Price is: ", rounded_discounted_price)
```

### Time Conversions

#### Convert total minutes to hours and minutes

```python
total_minutes = int(input("Enter the total minutes as an integer"))
hours = total_minutes//60
minutes= total_minutes%60

print(f'{total_minutes} minutes -> {hours} hours and {minutes} minutes')
```
