#### Check if number is odd and less than 10

```python
a = int(input("Enter the number: "))
print(a%2!=0 and a<10)
```

#### Check if number is odd and between -10 and 10 

```python
a = int(input("Enter the number: "))
print(a%2!=0 and -10<a<10)
```

#### Check if number has even no of digits but not more than 10 

```python
#catch
a = int(input("Enter the number: "))
print(len(str(abs(a))) % 2 ==0 and len(str(abs(a))) <=10)
```

### Discount Comparisons

```python
price1 = int(input("Enter price 1: "))
discount1 = float(input("Enter discount 1: "))
price2 = int(input("Enter price 2: "))
discount2 = float(input("Enter discount 2: "))

offer1 = price1*(1-discount1/100)
offer2 = price2*(1-discount2/100)

print("Is offer1 strictly cheaper [Note: This is a comparison on prices with different discounts]?")
print(offer1<offer2)
```
